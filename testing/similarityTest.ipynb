```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
f = open("image_results.json")
data = f.read()
```

```python
import json

data = json.loads(data)
```

```python
texts = [row["result"]["description"] for row in data]
```

```python
embeds = model.encode(texts, normalize_embeddings=True)
```

```python
query = model.encode("cute orange cat", normalize_embeddings=True)
```

```python
simlarity = (embeds @ query)
```

```python
import numpy as np

x = np.argsort(simlarity)[::-1]
```

```python
from IPython.display import Image
```

```python
print([data[i] for i in x])
Image(filename='./images/' + data[x[0]]['filename'], width='250') 
```
